```python
df = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load("s3://health-guardian-zeeshan/raw/public-seed/sleep/sleep.csv")
)

display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import current_timestamp, col

bronze_df = (
    df
    .withColumn("_ingested_at", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))
)
```

```python
display(bronze_df)
```

```python
# Write that as delta

(
    bronze_df.write
    .format("delta")
    .mode("append")
    .saveAsTable("health_and_habit_guardian.bronze.sleep")
)
```
