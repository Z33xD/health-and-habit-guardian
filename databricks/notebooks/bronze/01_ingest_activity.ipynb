```python
df = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load("s3://health-guardian-zeeshan/raw/public-seed/activity/activity.csv")
)

display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import current_timestamp, col

bronze_df = (
    df
    .withColumn("_ingested_at", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))
)
```

```python
display(bronze_df)
```

```python
# Write that as delta

(
    bronze_df.write
    .format("delta")
    .mode("append")
    .saveAsTable("health_and_habit_guardian.bronze.activity")
)
```

```python
%sql
SELECT COUNT(*) AS row_count
FROM health_and_habit_guardian.bronze.activity;
```

```python
%sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(user_id) AS user_ids,
    COUNT(date) AS dates,
    COUNT(steps) AS steps,
    COUNT(active_minutes) AS active_minutes
FROM health_and_habit_guardian.bronze.activity;
```

```python
%sql
SELECT
    user_id,
    date,
    COUNT(*) AS count
FROM health_and_habit_guardian.bronze.activity
GROUP BY user_id, date
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    MIN(steps) AS min_steps,
    MAX(steps) AS max_steps,
    MIN(resting_hr) AS min_resting_hr,
    MAX(resting_hr) AS max_resting_hr
FROM health_and_habit_guardian.bronze.activity;
```
